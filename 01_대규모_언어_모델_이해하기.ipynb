## Chapter 1 개요 <a class="anchor" id="chapter1"></a>
1. 오픈AI의 ChatGPT와 같은 대규모 언어 모델은 지난 몇 년간 발전한 심층 신경망(Deep Neural Network, DNN)이다.
   - 자연어 처리(Natural Language Processing, NLP) 분야에서 혁신적인 성과를 이루어냈다.

2. LLM(대규모 언어 모델, Large Language Model)이 사람의 언어를 이해하고, 생성하고, 해석하는 놀라운 능력을 가졌다.
    - 하지만 사람과 같은 의식이나 이해력을 가지고 있다는 뜻은 아니다.

3. 대량의 텍스트 데이터에서 훈련되기 때문에 사람의 언어에 있는 미묘한 문맥 정보를 깊이 이해할 수 있다.
    - 텍스트 번역, 감성 분석, 질문 답변등을 포함한 광범위한 NLP에서 작성 성능이 크게 향상되었다.

4. NLP는 일반적으로 텍스트 분류, 언어 번역등과 같은 특정 작업에 맞도록 고안되었다.
    - 반면에 LLM은 다양한 NLP 작업을 수행할 수 있는 범용 모델이다.

5. LLM의 성공은 많은 LLM의 기반이 되는 트랜스포머 구조와 방대한 훈련 데이터 덕분이다.
    - 수동으로 인코딩하기 어려운 다양한 언어적 뉘앙스, 문맥, 패턴을 감지할 수 있다.


## Chapter 2 LLM이란 무엇인가요? <a class="anchor" id="chapter2"></a>
1. LLM(대규모 언어 모델, Large Language Model)은 사람의 텍스트를 이해하고, 응답하다록 고안된 신경망(Neural Network)이다.
   - 대용량의 텍스트 데이터에서 훈련된 심층 신경망(Deep Neural Network, DNN)이다.
   - 인터넷에 공객된 전체 텍스트의 상당 부분을 사용해 훈련된다.
   - 신경망은 하나의 거대한 수학 함수이며 파라미터는 이런 수학 함수의 계수로 표현할 수 있다.

2. '대규모 언어 모델'에서 '대규모'는 모델의 크기와 훈련 데이터의 양을 의미한다.
   - 수십억 개의 파라미터를 가지고 있고, 파라미터는 시퀸스의 다음 단어를 예측하는 훈련에서 사용되는 신경망의 가중치이다.
   - 조작할 수 있다는 의미로 노브(Knob) 또는 다이얼이라고도 부르기도한다.

3. 다음 단어 예측(Next Token Prediction)은 언어의 고유한 순차 특징을 사용하여 텍스트 안의 맥락, 구조, 관계를 이해하는 모델을 훈련하는 방법이다.
   - 예를 들어, "The cat is on the"라는 문장이 주어지면, 모델은 "mat"이라는 단어를 예측할 수 있다.

4. LLM은 예측을 수행할 때 입력의 단어 중 어떤 단어에 선택적으로 주의를 기울일 수 있는 트랜스포머(Transformer)라는 신경망 구조를 사용한다.
   - 사람의 언어에 있는 뉘앙스와 문맥을 이해하는 데 도움이 된다.

5. LLM은 텍스트를 생성할 수 있기 때문에 생성형 인공지능(Generative AI)으로 분류된다.
   - LLM은 딥러닝(Deep Learning) 기술을 활용한 특정 애플리케이션이다.
   - 딥러닝은 다층 신경망(Deep Neural Network, DNN)을 사용하여 복잡한 패턴과 관계를 학습하는 머신러닝(Machine Learning)의 한 분야이다.

      ![예측](image/01-01-LLM.png)

6. 머신러닝은 AI 구현에 사용되는 알고리즘을 개발하는 데 초점을 맞추고있다.
   - 명시적으로 프로그래밍하지 않아도 데이터에서 학습하고 이를 기반으로 예측하는 알고리즘을 개발하는 것이다.
   - 스팸 메일을 찾아내기 위해 수동으로 규칙을 작성하는 대신, 머신러닝 알고리즘은 스팸과 정상 메일의 예시를 분석하여 스팸을 식별하는 패턴을 학습할 수 있다.
   - 훈련 데이터셋에 있는 샘플에 대한 예측 오류를 최소화 하면서 스팸을 특정하는 패턴과 특징을 인식하도록 학습된다.
   - 이메일 텍스트에서 특정단어 '경품', '무료'와 같은 특성을 수동으로 추출하고, 전문가가 정의한 특성을 기반으로 만든 데이터 셋을 사용해 모델을 훈련한다.

7. 딥러닝은 머신러닝의 하위 분야로, 3개 이상의 층을 가진 다층 신경망(Deep Neural Network, DNN)을 사용하여 복잡한 패턴과 관계를 학습한다.
   - 딥러닝은 이미지 인식, 음성 인식, 자연어 처리등의 분야에서 뛰어난 성능을 보여주고 있다.
   - 딥러닝 모델은 원시 데이터에서 특징을 자동으로 추출할 수 있기 때문에, 전통적인 머신러닝 알고리즘보다 더 복잡한 문제를 해결할 수 있다.

## Chapter 3 LLM 애플리케이션 <a class="anchor" id="chapter3"></a>
1. LLM은 비정형 텍스트 데이터(unstructured Text Data)를 처리하는 데 뛰어난 능력을 가지고 있다.
   - 비정형 텍스트 데이터는 구조화되지 않은 자연어 텍스트로, 이메일, 소셜 미디어 게시물, 기사, 리뷰등이 포함된다.
   - LLM은 이러한 데이터를 분석하고 이해하여 유용한 정보를 추출할 수 있다.
   - 감성 분석, 텍스트 요약 그 외 많은 작업에 사용되고 있고, 최근에는 소설, 기사, 컴퓨터 코드와 같은 텍스트를 생성하는 데에도 사용되고 있다.
   - 구글과 같은 검색엔진 대신에 사용되기도 한다.

2. LLM을 사용하여 오픈 AI의 ChatGPT나 구글의 Gemini와 같은 챗봇과 가상 비서를 구현할 수 있다.
   - 이러한 시스템은 자연어로 된 질문과 명령을 이해하고, 인간과 유사한 방식으로 응답할 수 있다.
   - 고객 지원, 개인 비서, 교육 도우미등 다양한 분야에서 활용되고 있다.

3. 의료나 법률과 같은 전문 영역의 대용량 텍스트에서 지식을 검색하기 위한 효과적인 도구로 사용할 수 있다.
   - LLM은 복잡한 질문을 이해하고, 관련 정보를 추출하며, 명확하고 간결한 답변을 생성할 수 있다.
   - 의료 기록 분석, 법률 문서 검토, 연구 논문 요약등에 활용되고 있다.

4. 주어진 이미지를 이해하고 응답에 참고할 수 있다.
    - 예를 들어, 사용자가 이미지에 대해 질문하면, LLM은 이미지의 내용을 분석하고, 관련 정보를 제공할 수 있다.
       - 이미지를 캡션과 같이 텍스트로 분석한다.
    - 시각적 질문 응답, 이미지 캡션 생성, 이미지 기반 검색등에 활용되고 있다.

5. LLM은 텍스트 분석과 생성에 관련된 거의 모든 작업을 자동화 하는 데 매우 유용하다.
    - 고객 지원, 콘텐츠 생성, 데이터 분석등 다양한 분야에서 활용되고 있다.
    - LLM은 자연어 처리 분야에서 혁신적인 도구로 자리잡고 있으며, 앞으로도 그 활용 범위가 계속해서 확장될 것으로 기대된다.
    - 대화를 활용해 기술을 직관적이며 사용하기 쉽게 만드는 잠재력이있다.

## Chapter 4 LLM의 구축 단계 <a class="anchor" id="chapter4"></a>
1. LMM을 구현해 보면 LLM 매커니즘과 한계를 이해하는데 크게 도움이 된다.
    - 기존의 오픈 소스 LLM을 특정 도메인의 데이터셋이나 작업에 적용하기 위해 사전 훈련하거나 미세 튜닝하는 데 필요한 지식을 습득할 수 있다.

2. 연구에 따르면, 모델링 성능 측면에서 특정 작업이나 도메인에 특화된 사용자 정의 LLM이 ChatGPT와 같은 범용 LLM보다 더 나은 성능을 보일 수 있다.
    - 특정 도메인에 특화된 LLM은 해당 도메인의 전문 용어, 문맥, 요구사항을 더 잘 이해할 수 있기 때문이다.
    - 예를 들어, 의료 분야에 특화된 LLM은 의료 기록, 연구 논문, 임상 지침등의 텍스트를 분석하고 이해하는 데 더 뛰어난 성능을 보일 수 있다.

4. 사용자 정의 LLM은 특히 데이터 개인정보 보호 측면에서 몇 가지 장점이 있다.
    - 기업은 기밀을 유지하기 위해 오픈AI와 같은 서드파티 LLM 제공 업체와 민감한 데이터를 공유하고 싶지 않을 수 있다.
    - 작은 규모의 LLM은 랩톱이나 스마트폰과 같은 사용자 장치에 바로 배포할 수 있다.
       - 로컬 구현은 응답 대기 시간을 크게 줄이고 서버 관련 비용을 절감할 수 있다.

5. LLM을 개발하는 일반적인 과정은 사전 훈련과 미세 튜닝으로 구성된다.
    - 사전 훈련의 '사전'은 LLM 같은 모델을 다양한 대규모 데이터셋에서 훈련시켜 언어에 대한 폭넓은 이해를 쌓는 초기 단계를 의미한다.
        - 사전 훈련된 모델을 파운데이션 모델로 사용한다.
        - 레이블이 없는 텍스트에서 다음 단어를 예측하면서 훈련을 진행한다.
    - 미세 튜닝은 특정 작업이나 도메인에 맞게 모델을 조정하는 과정이다.
       - 이 단계에서는 모델이 특정 작업이나 도메인의 요구사항을 더 잘 이해할 수 있도록 도와준다.
       - 모델을 구체적인 작업이나 도메인을 위한 특정 데이터셋에서 훈련한다.

        ![구축단계](image/01-03-process2.png)

6. LLM을 만드는 첫 번째 단계는 원시 텍스트(raw text)라고 부르는 대규모 텍스트 말뭉치(corpus)에서 훈련하는 것이다.
   - '원시'란 데이터에 레이블 정보가 없는 일반적인 텍스트라는 의미이다.
   - 서식 문자나 알 수 없는 언어로 된 문서를 제거하는 등의 필터를 적용할 수 있다.

7. LLM의 첫 번째 훈련 단계를 사전 훈련(Pretraining)이라고 부른다.
    - 사전 훈련된 LLM을 베이스 모델(Base Model) 또는 파운데이션 모델(Foundation Model)이라고 부른다.
    - 대표적인 예는 ChatGPT-3로 사용자가 일부 쓰다 만 문장을 완성시키는 텍스트 완성 능력이 있다.
        - 몇 개의 샘플을 기반으로 새로운 작업을 수행하는 방법을 배우는 퓨-샷 학습(Few-Shot Learning)을 제한적으로 수행할 수 있다.

8. 사전 훈련된 LLM을 얻은 후 레이블이 있는 데이터에서 LLM을 추가 훈련하는 미세튜닝을 수행할 수 있다.
    - 가장 인기 있는 미세 튜닝 LLM의 종류는 지시 미세튜닝(Instruction Fine-Tuning)과 분류 미세튜닝(Classification Fine-Tuning)이다.
       - 지시 미세 튜닝에서는 지시와 정답 쌍으로 데이터 셋(텍스트를 번역하기 위한 쿼리와 번역된 텍스트)이 구성된다.
       - 분류 미세 튜닝에서는 텍스트와 클래스 레이블로 데이터 셋(이메일과 스팸/비스팸 레이블)이 구성된다.

9. 머신러닝 분야는 공개된 정보가 풍부하다.
   - 많은 LLM이 오픈 소스(오픈 웨이트(Open Weights))로 제공되고 있다.
      - 훈련 데이터나, 훈련 방법은 공개가 되지않은 경우가 많이있다.

## Chapter 5 트랜스포머 구조 소개 <a class="anchor" id="chapter5"></a>
1. 대부분의 최신 LLM은 2017년 구글에서 발표한 논문 "Attention is All You Need"에서 처음 소개된 신경망 구조인 트랜스포머(Transformer)를 기반으로 하고 있다.
    - LLM을 이해하려면 영어 텍스트를 독일어와 프랑스어로 번역하는 기계 번역을 위해 개발된 원본 트랜스포머을 알아야한다.

        ![트랜스포머](image/01-04-transformer2.png)

2. 인코더 모듈은 입력 텍스트를 처리한다.
    -  입력의 문맥 정보를 포착하여 일련의 수치 표현 또는 백터로 인코딩한다.

3. 디코더 모듈이 인코딩된 벡터를 받아 출력 텍스트를 생성한다.
   - 디코더는 인코딩된 정보를 기반으로 다음 단어를 예측하고, 이를 반복하여 최종 출력을 생성한다.

4. 인코더와 디코더 모두 셀프 어텐션 메커님즘으로 연결된 많은 층으로 구성되어 있다.
   - 셀프 어텐션 메커니즘은 입력 시퀀스의 각 단어가 시퀀스의 다른 단어에 주의를 기울일 수(가중치를 부여할 수 있도록) 있도록 한다.
   - 이를 통해 모델은 문맥 정보를 더 잘 이해하고, 더 정확한 번역을 생성할 수 있다.

5. GPT는 생성  작업을 위해 고안되었지만 BERT 및 유사 모델들은 주어진 문장에서 마스킹되거나 가려진 단어를 예측하는 마스킹된 단어 예측에 특화되어있다.
   - BERT는 양방향 트랜스포머 인코더를 사용하여 문맥 정보를 포착한다.
   - BERT는 문장 분류, 감성 분석, 질문 답변등의 작업에 널리 사용되고 있다.

        ![BERT](image/01-04-bert.png)

6. GPT는 원본 트랜스포머 구조의 디코더 부분만 사용하면서 텍스트 생성이 필요한 작업을 위해 고안되었다.
   - GPT는 단방향 트랜스포머 디코더를 사용하여 다음 단어를 예측한다.
   - GPT는 텍스트 완성, 대화 생성, 이야기 생성등의 작업에 널리 사용되고 있다.
   - 제로-샷 학습과 퓨-샷 학습을 지원한다.
        - 제로-샷 학습(Zero-Shot Learning)는 사전에 어떤 구체적인 예시도 없이 처음 보는 작업에 일반화 하는 능력이다.
        - 퓨-샷 학습(Few-Shot Learning)은 몇 개의 예시만으로 새로운 작업을 수행하는 능력이다.

        ![GPT](image/01-04-gpt.png)

7. 오늘날의 LLM은 트랜스포머 구조를 기반으로 하기에 트랜스포머와 LLM은 동의어로 사용되는 경우가 많이 있다.
    - 하지만 모든 트랜스포머가 LLM은 아니다.
    - 합성곱 신경망 기반 LLM도 있기 때문에 모든 LLM이 트랜스포머인 것은 아니다.


## Chapter 6 대규모 데이터셋 활용하기 <a class="anchor" id="chapter6"></a>
1. GTP 및 BERT 유사 모델의 대규모 훈련 데이터셋은 수십억 개의 단어가 들어가 있는 텍스트 말뭉치로 구성된다.
    - 다양한 주제와 자연어, 컴퓨터 언어가 포함되어 있다.
    - 토큰은 모델이 텍스트를 읽는 단위
    - 훈련 데이터넷의 크기와 다양성으로 인해 모델이 언어 구문, 문법, 맥락에 과련된 다양한 작업이 가능하다.
    - 심지어 창의적인 작업(예: 시, 이야기 작성)도 가능하다.

    ![GPT3](image/01-05-GPT3.png)

2. 이런 모델의 사전 훈련 특성으로 인해 매우 다양한 후속 작업을 위해 미세 튜닝할 수 있다.

3. LLM을 사전 훈련하려면 상당한 자원과 비용이 필요하다.
    - 수십억 개의 매개변수를 가진 모델을 훈련하려면 수천 개의 GPU와 수주 또는 수개월이 걸릴 수 있다.
    - 따라서 대부분의 조직은 오픈 소스 LLM을 사용하거나, 상용 LLM API를 사용한다.

4. 사전 훈련된 LLM이 오픈 소스로 많이 공개되어 있다.
    - 예를 들어, GPT-2, GPT-3, BERT, RoBERTa, T5등이 있다.
    - 이러한 모델은 Hugging Face와 같은 플랫폼에서 쉽게 접근할 수 있다.
    - 오픈 소스 LLM을 사용하면 모델을 처음부터 훈련하는 데 필요한 시간과 비용을 절약할 수 있다.
    - 이를 범용 도구로 사용해 훈련 데이터에 없는 텍스트를 작성하고, 검색하고, 편집할 수 있다.

## Chapter 7 GPT 구조 자세히 살펴보기 <a class="anchor" id="chapter7"></a>
1. GPT는 오픈AI의 알렉 래드포트 등인 쓴 "Improving Language Understanding by Generative Pre-Training" 논문에서 처음 소개되었다.
   - GPT-3은 이 모델의 확장 버전으로 파라미터가 더 많고, 더 큰 데이터셋에서 훈련되었다.
   - 오픈AI의 InstructGPT 논문(https://arxiv.org/abs/2203.02155)에 나온 방법으로 대규모 지시 데이터 셋에서 미세 튜닝되었다.
   - 텍스트 완성에 유능하며 맞춤법 수정, 분류, 언어 번역과 같은 다른 작업을 수행할 수 있다.

2. 다음단어 예측 작업은 셀프 레이블링(self-labeling)의 하나인 자기 지도 학습의 한 형태이다.
    - 셀프 레이블링은 모델이 입력 데이터에서 스스로 레이블을 생성하는 방법이다.
    - 예를 들어, "The cat is on the mat"라는 문장에서 "mat"이라는 단어를 가리면, 모델은 나머지 단어를 기반으로 "mat"을 예측하도록 훈련된다.
       - 이 예에서 "mat"은 모델이 예측해야하는 레이블로 사용할 수 있다.
    - 이 방법은 대규모 레이블이 없는 텍스트 데이터에서 모델을 훈련하는 데 매우 효과적이다.
    - 레이블을 명시적으로 수집할 필요가 없고 데이터 구조 자체를 활용할 수 있다.

3. GPT 모델은 트랜스포머의 인코더 없이 디코더 모듈로만 구성된다.
    - 한 번에 한 단어씩 예측하여 텍스트를 생성하기 때문에 자기회귀 모델(Autoregressive Model)의 한 유형으로 간주된다.
    - 자기회귀 모델은 이전에 생성된 단어를 기반으로 다음 단어를 예측한다.
    - GPT 모델에서는 이전 시퀸스를 기반으로 다음 단어를 선택하는 식으로 출력 텍스트의 일관성을 향상시킨다.
    - 왼쪽에서 오른쪽 단방향 처리를 위해 고안되었으므로 한 번에 한 단어씩 반복적인 형태로 텍스트를 생성 작업에 적합하다.

        ![GPT3](image/01-06-GPT3-2.png)

4. 2020년에 출시된 GPT-3는 딥러닝과 대규모 언어 모델의 개발 수준으로 볼 때 구식 모델로 간주된다.
    - 하지만 메타의 Llma, 오픈AI의 GPT-4, 구글의 Gemini와 같은 최신 모델은 GPT-3의 기본 아키텍처를 기반으로 한다.

5. 모델이 명시적으로 훈련되지 않은 작업을 수행하는 능력을 창발적 행동(Emergent Behavior)이라고 부른다.
    - 예를 들어, GPT-3는 번역, 질문 답변, 요약과 같은 작업을 수행할 수 있다.
       - 모델이 번역을 위해 훈련되어 있지 않더라도 언어 사이의 번역 패턴을 학습하고 번역 작업을 할 수 있다.
    - 이러한 작업은 모델이 명시적으로 훈련되지 않은 작업이다.
    - 모델이 훈련 데이터에서 언어의 패턴과 구조를 학습했기 때문에 이러한 작업을 수행할 수 있다.
    - 작업마다 개별 모델을 사용할 필요 없이 하나의 모델로 다양한 작업을 할 수 있다.
