### Chapter 1 개요 <a class="anchor" id="chapter1"></a>
1. 오픈AI의 ChatGPT와 같은 대규모 언어 모델은 지난 몇 년간 발전한 심층 신경망(Deep Neural Network, DNN)이다.
   - 자연어 처리(Natural Language Processing, NLP) 분야에서 혁신적인 성과를 이루어냈다.

2. LLM(대규모 언어 모델, Large Language Model)이 사람의 언어를 이해하고, 생성하고, 해석하는 놀라운 능력을 가졌다.
    - 하지만 사람과 같은 의식이나 이해력을 가지고 있다는 뜻은 아니다.

3. 대량의 텍스트 데이터에서 훈련되기 때문에 사람의 언어에 있는 미묘한 문맥 정보를 깊이 이해할 수 있다.
    - 텍스트 번역, 감성 분석, 질문 답변등을 포함한 광범위한 NLP에서 작성 성능이 크게 향상되었다.

4. NLP는 일반적으로 텍스트 분류, 언어 번역등과 같은 특정 작업에 맞도록 고안되었다.
    - 반면에 LLM은 다양한 NLP 작업을 수행할 수 있는 범용 모델이다.

5. LLM의 성공은 많은 LLM의 기반이 되는 트랜스포머 구조와 방대한 훈련 데이터 덕분이다.
    - 수동으로 인코딩하기 어려운 다양한 언어적 뉘앙스, 문맥, 패턴을 감지할 수 있다.


### Chapter 2 LLM이란 무엇인가요? <a class="anchor" id="chapter2"></a>
1. LLM(대규모 언어 모델, Large Language Model)은 사람의 텍스트를 이해하고, 응답하다록 고안된 신경망(Neural Network)이다.
   - 대용향의 텍스트 데이터에서 훈련된 심층 신경망(Deep Neural Network, DNN)이다.
   - 인터넷에 공객된 전체 텍스트의 상당 부분을 사용해 훈련된다.

2. '대규모 언어 모델'에서 '대규모'는 모델의 크기와 훈련 데이터의 양을 의미한다.
   - 수십억 개의 파라미터를 가지고 있고, 파라미터는 시퀸스의 다음 단어를 예측하는 훈련에서 사용되는 신경망의 가중치이다.

3. 다음 단어 예측(Next Token Prediction)은 언어의 고유한 순차 특징을 사용하여 텍스트 안의 맥락, 구조, 관계를 이해하는 모델을 훈련하는 방법이다.
   - 예를 들어, "The cat is on the"라는 문장이 주어지면, 모델은 "mat"이라는 단어를 예측할 수 있다.

4. LLM은 예측을 수행할 때 입력의 단어 중 어떤 단어에 선택적으로 주의를 기울일 수 있는 트랜스포머(Transformer)라는 신경망 구조를 사용한다.
   - 사람의 언어에 있는 뉘앙스와 문맥을 이해하는 데 도움이 된다.

5. LLM은 텍스트를 생성할 수 있기 때문에 생성형 인공지능(Generative AI)으로 분류된다.
   - LLM은 딥러닝(Deep Learning) 기술을 활용한 특정 애플리케이션이다.
   - 딥러닝은 다층 신경망(Deep Neural Network, DNN)을 사용하여 복잡한 패턴과 관계를 학습하는 머신러닝(Machine Learning)의 한 분야이다.

      ![LLM](image/01-01-LLM.png)

6. 머신러닝은 AI 구현에 사용되는 알고리즘을 개발하는 데 초점을 맞추고있다.
   - 명식적으로 프로그래밍하지 않아도 데이터에서 학습하고 이를 기반으로 예측하는 알고리즘을 개발하는 것이다.
   - 스팸 메일을 찾아내기 위해 수동으로 규칙을 작성하는 대신, 머신러닝 알고리즘은 스팸과 정상 메일의 예시를 분석하여 스팸을 식별하는 패턴을 학습할 수 있다.
   - 훈련 데이터셋에 있는 샘플에 대한 예측 오류를 최소화 하면서 스팸을 특정하는 패턴과 특징을 인식하도록 학습된다.
   - 이메일 텍스트에서 특정단어 '경품', '무료'와 같은 특성을 수동으로 추출하고, 전문가가 정의한 특성을 기반으로 만든 데이터 셋을 사용해 모델을 훈련한다.

7. 딥러닝은 머신러닝의 하위 분야로, 3개 이상의 층을 가진 다층 신경망(Deep Neural Network, DNN)을 사용하여 복잡한 패턴과 관계를 학습한다.
   - 딥러닝은 이미지 인식, 음성 인식, 자연어 처리등의 분야에서 뛰어난 성능을 보여주고 있다.
   - 딥러닝 모델은 원시 데이터에서 특징을 자동으로 추출할 수 있기 때문에, 전통적인 머신러닝 알고리즘보다 더 복잡한 문제를 해결할 수 있다.
