## Chapter 2-1 긴 시퀸스 모델링의 문제점 <a class="anchor" id="chapter2_1"></a>
1. 텍스트를 한 언어에서 다른 언어로 번역할 때 생기는 주요 문제는 무엇일까요? 단순하게 한단어씩 번역하지 못하는 이유는 무엇인가요?
   - 소스 언어와 타켓 언어의 문법 구조가 다르기 때문에 한 단어씩 번역할 수 없다.
   - 어떤 단더는 소스 문장에서 앞 단어나 뒷 단어를 참조해서 번역해야한다.

2. 언어 번역을 위한 RNN 모델의 인코더와 디코더의 역활을 설명하세요.
    - 인코더
       - 입력데이터를 받아 순차적으로 처리한다.
       - 각 단계마다 은닉상태를 업데으트 하며, 전체 입력 텍스트르 하나의 은닉 상태 벡터로 압축한다.
    - 디코더
       - 매 스텝마다 은닉상태를 업데이트 하며, 이를 통해 다음 단어 예측에 필요한 문맥 정보를 다음 스텝에 전달한다.
       - 이전에 생성된 단어와 인코더의 최종 은닉 상태를 사용하여 다음 단어를 예측한다.

3. 긴 시퀸스를 처리하는 데 인코더-디코더 RNN의 주요 제약은 무엇인가요?
   - 최종 은닉 상태에만 의존하기 때문에 긴 시퀸스의 모든 정보를 효과적으로 캡처하지 못한다.
   - 멀리 떨어진 단어에 의존성이 있는 경우 이를 포착하기 어렵다.

4. 인코더-디코더 RNN 측면에서 은닉 상태의 개념을 설명하세요.
    - 입력 시퀸스의 압축된 표현으로 전체 텍스트의 의미를 포착한다.
    - 디코더가 번역된 출력을 생성하기 위해 사용하는 메모리 셀의 역할을 한다.

## Chapter 2-2 어텐션 메커니즘으로 데이터 의존성 포착하기 <a class="anchor" id="chapter2_2"></a>
1. 긴 텍스트를 번역하는데 RNN의 주요 제약은 무엇인가요?

    - 이를 통해 긴 시퀸스에서도 중요한 정보에 집중할 수 있어 번역 품질이 향상된다.

## Chapter 2-3 셀프 어텐션으로 입력의 서로 다른 부분에 주의 기울이기 <a class="anchor" id="chapter2_3"></a>
1. 셀프 어텐션의 '셀프'의 의미는 무엇인가요?
    - 하나의 입력 시퀸스 내의 원소들이 서로에게 주의를 기울인다는 의미이다.
    - 하나의 입력 시퀸스에 있는 서로 다른 위치의 원소 사이에서 어텐션 가중치를 계산한다.

2. 셀프 어텐션에서 문맥 벡터의 개념을 설명하세요.
    - 입력 시퀸스에 있는 모든 원소의 정보를 통합한 고품질 임베딩 벡터
    - 다른 원소와의 관계를 고려하여 각 원소를 이해한 결과물이다.

3. 셀프 어텐션에서 어텐션 점수의 역활은 무엇인가요?
    - 입력 시퀸스에 있는 서로 다른 원소사이의 관련성을 측정한다.
    - 어텐션 점수는 각 원소 사이의 유사성을 나타내며, 이를 통해 중요한 요소에 더 많은 주의를 기울일 수 있다.

4. 어텐션 점수에 정규화를 적용하는 이유는 무엇인가요?
    - 어텐션 점수를 확률 분포로 변환하여 각 원소의 상대적 중요성을 나타내기 위해서이다.
    - 정규화를 통해 어텐션 가중치가 0과 1 사이의 값이 되도록 하여 해석하기 쉽게 만든다.

## Chapter 2-4 훈련 가능한 가중치를 가진 셀프 어텐션 구현하기 <a class="anchor" id="chapter2_4"></a>
1. 셀프 어텐션 메커니즘에서 가중치 행렬 W<sub>q</sub>, W<sub>k</sub> , W<sub>v</sub>의 목적을 설명하시요
   - W_q: 입력 임베딩을 Query 공간으로 선형투영. 다른 토큰들을 '질문'할 특징을 만듭니다.
   - W_k: 입력 임베딩을 Key 공간으로 선형투영. 각 토큰의 '응답 가능한' 특징을 만듭니다.
   - W_v: 입력 임베딩을 Value 공간으로 선형투영. 최종 문맥(출력)으로 전달될 정보입니다.
   - 이 세 행렬 덕분에 모델은 어떤 특징으로 비교(유사도)할지(Q vs K)와, 비교 결과로 어떤 정보를 모아야 할지(V)를 학습할 수 있습니다.

2. 셀프 어텐션 메커니즘에서 어텐션 점수와 어텐션 가중치를 계산하는 과정을 설명하세요.
   - 훈련 가능한 가중치 행령 w_q, W_k, W_v를 생성한다.
   - 입력 벡터를 각각 쿼리, 키, 값 벡터로 선형 투형한다.
   - 입력 쿼리 벡터와 키 벡터 간의 행렬곱을 수행하여 어텐션 점수를 계산한다.
   - 소프트맥스 함수를 사용하여 어텐션 점수를 정규화하여 어텐션 가중치를 얻는다.
   - 값 벡터와 어텐션 가중치를 곱하여 문맥 벡터를 계산한다.

3. 스케일드 점곱 어텐션 메커니즘에서 임베딩 차원(d_k)의 제곱근으로 나누는 것이 왜 중요한가요?
   - 역전파에서 그레이디언트가 작아지는 것을 막는데 도움이 된다. 특히 임베딩 차원이 클 때 그렇다.
   - 더 안정적인 범위에서 소프트맥스 함수가 동작하게 되며 결국 더 효율적인 모델 훈련으로 이루어 집니다.

5. SelfAttention_v1과 SelfAttention_v2 클래스의 주요 차이점은 무엇인가요?
   - SelfAttention_v1은 가중치 행렬을 수동으로 초기화, SelfAttention_v2는 nn.Linear 레이어를 사용하여 가중치를 초기화한다.
   - Liner 층은 최적화된 가중치 초기화를 제공하고 훈련 안정성을 향상시기킨다.

## Chapter 2-5 코잘 어텐션으로 미래의 단어를 감추기 <a class="anchor" id="chapter2_5"></a>
1. 코잘 어텐션의 목적은 무엇인가요? 표준적인 셀프 어텐션과는 어떻게 다른가요?
   - 주어진 토큰으로 어텐션 점수를 계산할 때 시퀸스 이전의 입력과 현제 입력만 참조하도록 한다.
   - 각 토큰을 처리할 때 입력 토큰에서 현재 토큰의 다음에 오는 미래 토큰을 마스킹한다.
   - 표준적인 셀프 어텐션은 모든 토큰에 대해 어텐션을 계산한다.

2. 다음 그림은 무엇을 보여주나요?
   - 주대각선 위의 어텐션 가중치를 마스킹하고 마스킹하지 않은 어텐션 가중치를 정규화하여 각 행의 값이 1이 되도록한다.

      ![W-03-05-02](image/W-03-05-02.png)

3. 어텐션 가중치에 코잘 어텐션 마스크를 적용하는 과정을 설명하세요. 마스킹의 목적은 무엇인가요?
   - 순서
      - 주대각선 위의 어텐션 가중치를 음의 무한대로 마스킹한다.
      - 소프트맥스 함수를 사용하여 마스킹하지 않은 어텐션 가중치를 합이 1이 되도록 정규화한다.
   - 목적
      - 모델이 학습 하는 동안 미래의 단어를 참조하지 목하게한다.
      
4. 코잘 어텐션에서 정보 누수의 개념과 이를 해결하는 방법을 설명하세요.
   - 마스킹된 위치가 소프트맥스 계산에 영향을 미칠 때 일어납니다.
   - 마스킹한 후에 어텐션 가중치를 다시 정규화하면 마스킹된 위치의 영향을 없에므로 정보 누수를 방지할 수 있습니다.

5. 어텐션 메커니즘에서 드롭아웃의 목적은 무엇인가요? 코잘 어텐션에 어떻게 적용되나요?
   - 신경망이 특정 입력에 과도하게 의존하여 과적합되는 것을 방지한다.
   - 코잘어텐션의 어텐션 가중치에 드롭아웃을 적용하여 일부 어텐션을 무작위로 제거한다.
   - 드롭아웃 적용 후 남은 어텐션에 보상을 하기 위해 드롭아웃 비율만큼 어텐션 가중치를 스케일링한다.
   - torch.nn.Dropout 모듈을 사용하여 구현한다.

6. CausalAttention 클래스에서 register_buffer 메서드의 역활은 무엇인가요?
   - 메서드에 지정한 텐서를 모델과 함께 적절한 장치로 이동시키고 저장한다.
   - 훈련 과정에서 장치 불일치로 인한 오류를 방지한다.
