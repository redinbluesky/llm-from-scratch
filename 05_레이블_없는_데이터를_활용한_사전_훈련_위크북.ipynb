## Chapter 2-1 텍스트 생성 모델 평가하기 <a class="anchor" id="chapter2-1"></a>
1. Generate_text_simple 함수의 역활은 무엇이고 어떻게 동작하나요?
   - GPT 모델을 통하여 텍스트를 생성하는 함수이다. 
   - 토큰아이디 시퀸스를 입력 받아, 지정된 최대 길이까지 반복적으로 다음 토큰을 예측하고, 예측된 토큰을 시퀸스에 추가하여 새로운 텍스트를 생성한다.

2. 텍스트 생성 손실의 개념과 생성된 텍스트 품질을 평가하는 데 중요성을 설명하세요.
   - 목표로 하는 타킷 시퀸스와 모델이 생성한 시퀸스 간의 차이를 측정하는 지표이다.
   - 텍스트 손실 지표를 최소화하는 방향으로 모델을 훈련시키면, 모델이 더 정확하고 일관된 텍스트를 생성할 수 있다.

3. LLM 훈련에서 역전파의 역활과 이것이 텍스트 생성 손실과 어떤 관련이 있는지 설명하세요.   
   - 역전파는 출력층에서 입력층으로 가중치를 조정하는 과정이다.
   - 역전파를 통해 가중치를 조절하여 텍스트 생성 손실을 최소화하는 방향으로 모델을 최적화한다.

4. 크로스 엔트로피 손실이 무엇인가요? LLM 평가에 어떻게 사용되나요?
   - 크로스 엔트로피 손실은 실제 분포와 예측 분포 간의 차이를 측정하는 지표이다.
   - LLM 평가에서 모델이 생성한 토큰 확률 분포와 실제 타겟 토큰 분포 간의 차이를 측정하여 모델의 성능을 평가한다.

5. 다음 그림을 참고하여 번호와 설명을 올바르게 짝지으세요. 이탤릭체로 쓴 것은 이미 올바르게 짝지어진 것입니다.

6. 혼잡도의 개념과 크로스 엔트로피 손실과의 관계를 설명하세요.
   - 혼잡도는 모델이 얼마나 다음 토큰을 예측하는 데 어려움을 겪는지를 나타내는 지표이다.
