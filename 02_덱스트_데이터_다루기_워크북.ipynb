## Chapter 2-1 단어 임베딩 이해하기 <a class="anchor" id="chapter2_1"></a>
1. 딥러닝 모델에서 텍스트 데이터를 처리하는 데 단어 임베딩이 필요한 이유는 무엇인가요?
   - LLM을 포함한 신경망 모델은 텍스트 데이터를 이해할 수 없기 때문에 숫자형 데이터로 변환해야 한다.
   - 텍스트 단어를 벡터 형태로 변환하는 임베딩 작업이 필요한다.

2. 단어 임베딩을 생성하는 Word2Vec 방식의 핵심 아이디어는 무엇인가요?
   - 비슷한 맥락에 등장하는 단어는 비슷한 의미를 가지는 경향이 있다.
   - 단어를 2차원 임베딩 공간에 투영하면 비슷한 단어는 서로 가깝게 위치한다.

3. 단어 임베딩의 차원을 선택할 때 발생하는 트레이드오프를 설명하세요.
   - 차원이 너무 낮으면 단어의 의미를 충분히 표현하지 못할 수 있다.
   - 차원이 너무 높으면 과적합(overfitting) 위험이 증가하고 계산 비용이 높아진다.
   - 적절한 차원을 선택하는 것이 중요하다.

4. Word2Vec 같은 사전 훈련된 모델을 사용하는 것과 비교하여 LLM은 일반적으로 단어 임베딩을 어떻게 처리하나요?
   - LLM은 입력 층의 일부로 자체적인 임베딩을 만들고 훈련 과정에서 이를 최적화한다.
   - 사전 훈련된 임베딩을 사용하는 대신, 모델이 특정 작업에 맞게 임베딩을 학습한다.

5. 고차원 단어 임베딩을 시각화 하기 어려운 핵심 이유는 무엇인가요?
    - 인간이 인지할 수 있는 공간은 3차원까지이기 때문에, 고차원 데이터를 시각화하는 데 한계가 있다.
    - 차원 축소 기법(예: PCA, t-SNE)을 사용하여 2D 또는 3D로 변환해야 한다.

## Chapter 2-2 텍스트 토큰화 하기 <a class="anchor" id="chapter2_2"></a>
1. 대규모 언어 모델을 구축하는 데 텍스트 토큰화의 목적은 무엇인가요?
   - 토큰 임베딩을 생성하기 위해 텍스트를 토큰으로 분할한다.

2. 파이썬 정규 표현식 라이브러리 re를 사용하여 텍스트 토큰화 과정을 설명하세요.
   - 텍스트를 공백으로 분리한다.
   - 구두 점과 단어를 분리한다.
   - 공백을 제거한다.
   - 물음표, 따옴표, 이 대시와 같은 특수 문자를 처리한다.

3. LLM 훈련을 위해 텍스트를 토큰화할 때 대소문자를 고려하는 것이 왜 중요한가요?
    - 대소문자는 단어의 의미를 변경할 수 있다.
    - 예를 들어, "Apple"은 회사 이름이고 "apple"은 과일을 의미한다.
    - 따라서 대소문자를 구분하는 것이 중요하다.

4. 토큰화 과정에서 공객을 제거하는 거과 유지하는 것 사이에 트레이드오프를 설명하세요.
    - 공백을 제거하면 모델이 더 간결한 입력을 처리할 수 있지만, 문장 구조와 의미를 잃을 수 있다.
    - 공백을 유지하면 문장의 의미를 더 잘 보존할 수 있지만, 모델의 복잡성이 증가할 수 있다.
    - 특정 작업에 따라 적절한 접근 방식을 선택해야 한다.

## Chapter 2-3 토큰을 토큰 ID로 변환하기 <a class="anchor" id="chapter2_3"></a>
1. 토큰을 토큰 ID로 변환하는 목적은 무엇인가요?
   - 텍스트 데이터를 신경망이 처리할 수 있는 숫자형 데이터롤 변환한다.

2. 토큰화를 위해 어휘사전을 어떻게 만드나요?
   - 훈련 데이터셋에 있는 모든 개별 텍스트를 개별 토큰으로 분활 후 고유한 토큰 집합을 만든다.
   - 중복된 토큰을 제거하여 어휘사전을 생성한다.
   - 각 토큰에 고유한 정수 ID를 부여한다.

3. SimpleTokenizerV1 클래스에서 encode 메서드의 역활은 무엇인가요?
   - 입력 텍스트를 토큰화하고 각 토큰을 어휘사전에서 해당하는 토큰 ID로 변환한다.
   - 토큰 ID의 리스트를 반환한다.

4. SimpleTokenizerV1 클래스에서 decode 메서드의 역활은 무엇인가요?
   - 토큰 ID 리스트를 입력으로 받아 각 토큰 ID를 어휘사전에서 해당하는 토큰으로 변환한다.
   - 토큰들을 공백으로 연결하여 원래 텍스트 형태로 복원한다.

5. 작은 훈련 세트로 만든 어휘사전의 제약 사항은 무엇인가요?
   - 어휘사전이 작아서 없는 단어가 많을 수 있다.
   - 토큰화 과정에서 적절한 토큰이 생성되지 않을 수 있다.

## Chapter 2-4 특수 문맥 토큰 추가하기 <a class="anchor" id="chapter2_4"></a>
1. 어휘사전에 추가되는 두 개의 특수 토큰은 무엇이며 이 토큰의 역할은 무엇인가요?
   - <|endoftext|> : 관련이 없느 텍스트 사이에 추가되며, 모델이 문맥을 구별할 수 있도록 한다.
   - <|unk|>: 어휘사전에 없는 단어가 나타나면, 이 토큰으로 대체된다.

2. SimpleTokenizerV2 클래스에서 알지 못하는 단어는 어떻게 처리하나요?
   - 어휘사전에 없는 단어가 나타나면, 해당 단어를 <|unk|> 토큰으로 대체한다.

3. 여러 개의 독립된 문서에서 훈련할 때 <|endoftext|> 토큰의 역활을 설명하세요.
    - 관련이 없는 텍스트 사이에 <|endoftext|> 토큰을 추가하여 모델이 문맥을 구별할 수 있도록 한다.
    - 훈련을 위해 텍스트는 연결되어 있지만 관련은 없다고 알려준다.

4. 다음 코드는 일부 내용이 삭제되었습니다. 삭제된 코드는 무엇이며 어디에 들어가야 하나요?
   - c. <|unk|>

    ![W-02-04-04](image/W-02-04-04.png)

5. LLM에서 널리 사용되는 추가적인 특수 토큰은 무엇이며 어떤 기능을 하나요?
   - <|pad|>: 시퀀스 길이를 맞추기 위해 사용된다.
   - <|bos|>: 문장의 시작을 나타낸다.
   - <|eos|>: 문장의 끝을 나타낸다.
   - <|cls|>: 분류 작업에서 사용된다.

    - 알지 못하는 단어를 더 작은 부분 단위로 분할하여 어휘사전에 있는 토큰으로 매핑한다.

## Chapter 2-6 슬라이딩 윈도로 데이터 샘플링하기 <a class="anchor" id="chapter2_6"></a>

    - LLM은 다음 단어 예측 작업을 수행하기 때문에, 이전 토큰을 기반으로 다음 토큰을 예측하는 입력-타깃 쌍이 필요하다.

2. 입력-타깃 쌍을 생성하기 위해 사용하는 슬라이딩 윈도 방법과 작동 방식을 설명하세요.
    - 텍스트 데이터에서 고정 길이의 윈도우를 설정하고 윈도우를 스트라이드 만큼 이동 시키면서 입력-타깃 쌍을 생성한다.
