### Section 2-1 LLM이란 무엇인가요? <a class="anchor" id="section2-1"></a>
1. 대규모 언어 모델이란 무엇이고 어떻게 동작하나요?
   - 대규모 모델(LLM Large Language Model)은 사람의 텍스트를 이해하고, 응답하도록 고안된 심층 신경망입니다.
   - 입력의 다른 부분에 주의를 기울이기 위해 트랜스포머 구조를 사용하며 언어에 있는 뉘앙스를 처리하는데 능숙하다
   - 시퀀스에 있는 다음 단어를 예측하는 작업으로 훈련되며 이를 통해 텍스트 안에 있는 문맥, 구조, 관계를 학습합니다.

2. 대규모 언어 모델에서 대규모의 의미는 무엇인가요?
   - '대규모'는 모델의 크기와 훈련 데이터의 양을 의미합니다.
   - 수십억 개의 파라미터를 가지고 있고, 방대한 양의 텍스트 데이터로 훈련됩니다.

3. LLM과 생성 AI(Generative AI)는 어떻게 관련이 있나요?
    - LLM은 생성 AI의 한 유형으로, 텍스트를 생성하는 데 특화되어 있습니다.
    - 생성 AI는 이미지, 음악 등 다양한 형태의 콘텐츠를 생성할 수 있는 모델을 포함합니다.

4. AI와 머신러닝, 딥러닝의 차이점을 설명하세요.
    - AI는 사람 수준의 지능을 필요로 하는 작업이 가능한 기계를 만드는 광범위한 분야입니다.
    - 머신러닝은 AI의 하위 분야로 데이터에서 학습하는 알고리즘에 초점을 맞추고있다.
    - 딥러닝은 인공 신경망을 사용하여 복잡한 패턴을 학습하는 머신러닝의 하위 분야입니다.

5. 아래의 그림에서 각 번호에 해당하는 텍스트는 무엇인가요?
   1) LLM(Large Language Model)
   2) 생성형 AI(Generative AI)
   3) 머신러닝(Machine Learning)

        ![W-01-5](image/W-01-5.png)

6. 특성 추출 입장에서 전통적인 머신러닝과 딥러닝의 차이점은 무엇인가요?
    - 전통적인 머신러닝에서는 수동으로 특성을 추출하고 이를 기반으로 데이터셋을 구성한다.
       - 예: 이메일의 스팸 여부를 분류할 때, 특정 단어 '경품', '무료' 등을 수동으로 추출하여 특성으로 사용.
    - 딥러닝에서는 원시 데이터에서 특징을 자동으로 추출하여 모델에 입력한다.
       - 예: LLM이 문장을 받아서 다음 단어 예측을 할 때, 다음 단어를 특성으로 자동으로 추출.

### Section 2-2 LLM 애플리케이션 <a class="anchor" id="section2-2"></a>
1. 다양한 영역에 걸친 LLM의 주요 애플리케이션에는 어떤 것이 있나요?
    - 감성 분석, 텍스트 요약
    - 구글 검색 엔진 대신에 사용
    - 고객지원, 개인 비서, 교육 도우미
    - 의료와 법률과 같은 전문 분야에서 텍스트 검색

2. 챗봇과 가상 비서 개발에 LLM이 어떻게 기여하나요?
    - 자연어로된 질문을 이해하고 인간과 유사한 방식으로 응답한다.
    - 복잡한 질문을 이해하고, 관련된 정보를 추출하여 정확한 답변을 제공한다.
    - 대화의 맥락을 파악하여 사용자와 자연스러운 대화를 나눈다.

3. 전문 분야의 지식 검색에서 LLM의 역활을 설명하세요.
    - 의료나 법률과 같은 전문 분야에서 방대한 양의 텍스트 데이터를 처리하고 이해한다.
    - 복잡한 질문을 이해하고, 관련된 정보를 추출하여 정확한 답변을 제공한다.
    - 의료 기록 분석, 법률 문서 검토 등에서 중요한 정보를 신속하게 찾아낸다.

4. 기술과 사람의 관계에 대해 LLM이 미칠 잠재적인 영향은 무엇인가요?
    - 대화를 사용해 기술을 더 쉽게 사용할 수 있게한다.
    - 생산성을 높이고, 복잡한 작업을 자동화한다.
    - 교육과 학습을 지원하여 지식 접근성을 향상시킨다.

### Section 2-3 LLM 구축 단계 <a class="anchor" id="section2-3"></a>
1. ChatGPT와 같은 범용 목적의 LLM을 사용하는 것과 비교해 맞춤형 LLM을 구축하는 주요 이점은 무엇인가요?
    - 특정 작업이나 도메인에 대해 더 나은 성능을 발휘할 수 있다.
    - 특정 도메인에서 사용하는 전문용어나 문맥, 요구사항을 더 잘 이해할 수 있다.
    - 민감한 데이터나 독점 정보를 처리할 때 보안과 프라이버시를 강화할 수 있다.
    - 조직의 특정 요구사항에 맞게 모델을 조정할 수 있다.
    - 스마트폰과 같은 사용자 장치에 바로 배포할 수 있다.

2. LLM을 만들기 위한 두 단계 훈련 과정을 설명하세요.
    - 사전 훈련
      - 다양한 대규모 데이터 셋에서 훈련하여 언어에 대한 이해를 높인다.
      - 레이블이 없는 원시 텍스트에서 다음 단어를 예측하는 작업으로 훈련한다.
    - 미세 듀닝
       - 특정 작업이나 도메인에 맞게 모델을 조정한다.
       - 레이블이 있는 데이터 셋을 사용하여 모델을 훈련한다.

3. LLM 사전 훈련의 목적은 무엇인가요? 이 단계에서는 어떤 유형의 데이터가 사용되나요?
    - 목적: 언어에 대한 일반적인 이해를 가진 파운데이션 모델을 구축한다.
    - 데이터 유형: 레이블이 없는 원시 텍스트 데이터(예: 책, 기사, 웹사이트).

4. LLM 사전 훈련의 맥락에서 자기 지도 학습의 개념을 설명하세요.
   - 자기 지도 학습은 레이블이 없는 데이터에서 학습하는 방법이다.
   - 모델이 다음 단어를 예측하는 작업을 통해 언어의 구조와 패턴을 학습한다.

5. LLM 미세 튜닝의 두 가지 주요 방법은 무엇인가요? 레이블을 가진 데이터를 사용한다는 측면에서 두 방법이 어떻게 다른가요?
   - 지시 미세튜닝: 지시와 정답 쌍을 사용하여 모델을 훈련한다.
   - 분류 미세튜닝: 텍스트와 클래스 레이블 쌍을 사용하여 모델을 훈련한다.
   - 지시 미세튜닝은 특정 작업에 대한 지시를 제공하는 반면, 분류 미세튜닝은 텍스트를 특정 클래스에 할당하는 데 중점을 둔다.

### Section 2-4 트랜스포머 구조 소개 <a class="anchor" id="section2-4"></a>
1. 트랜스포머 구조가 무엇이고 LLM 개발에서 얼마나 중요한가요?
    - 입력 텍스트를 받는 인코더와 이를 받아 출력을 생성하는 디코더로 구성된 신경망 아키텍처입니다.
    - 현재 대부분의 LLM은 트랜스포머 구조를 기반으로 하며, 이는 모델이 긴 문맥을 이해하고 복잡한 언어 패턴을 학습하는 데 매우 효과적이기 때문입니다.

2. 다음 그림에서 임베딩 단계의 출력은 1, 2, 3 중 몇 번에 해당하나요? 
    - 2번

        ![W-01-04-02](image/W-01-04-02.png)

3. 트랜스포머 구조에 있는 두 개의 주요 구성 요소와 언어 처리애 대한 역활을 설명하세요.
    - 인코더: 입력 텍스트를 받아 이를 벡터로 인코딩한다. 
    - 디코더: 인코딩된 벡터를 기반으로 다음 단어를 예측하고, 이를 반복하여 출력을 생성한다.

4. 셀프 어텐션 메커니즘이 무엇인가요? 트랜스포머의 능력에 어떤 기여를 하나요?
    - 시퀸스에 있는 단어 간의 상대적 중요도에 가중치를 부여할 수 있다.
    - 이를 통해 모델은 문맥 정보를 더 잘 이해하고, 더 정확한 예측을 할 수 있다.

5. 훈련 방식과 주요 애플리케이션 맥락에서 BERT 모델과 GPT 모델 사이의 주요 차이점을 설명하세요.
    - BERT: 양방향 인코더 모델로, 문장의 양쪽 컨텍스트를 모두 고려하여 단어를 이해한다. 주로 텍스트 분류, 감성 분석 등에 사용된다.
    - GPT: 단방향 디코더 모델로, 이전 단어들만을 고려하여 다음 단어를 예측한다. 주로 텍스트 생성, 챗봇 등에 사용된다.

6. 제로-샷 학습과 퓻-샷 할습이 무엇인가요? GPT 모델과 어떤 관련이 있나요?
    - 제로-샷 학습: 모델이 특정 작업에 대해 훈련되지 않았음에도 불구하고, 해당 작업을 수행할 수 있는 능력이다.
    - 퓻-샷 학습: 모델이 소량의 예시만으로 새로운 작업을 수행할 수 있는 능력이다.
    - GPT 모델은 대규모 사전 훈련 덕분에 제로-샷 및 퓻-샷 학습 능력을 갖추고 있다.

### Section 2-5 대규모 데이터셋 활용하기 <a class="anchor" id="section2-5"></a>
1. GPT-3와 BERT 같은 대규모 언어 모델에서 사용되는 훈련 데이터셋의 주요 특징은 무엇인가요?
   - 수십억개의 단어가 들어가 있는 텍스트 말뭉치로 구성된다.
   - 다양한 주제와 자연어, 컴퓨터 언어가 포함되어 있다.
   - 이런 다양성으로 인해 모델이 언어의 구문, 문법, 맥락에 대한 깊은 이해를 발전시킬 수 있다.

2. 대규모 언어 모델의 성능을 위해 훈련 데이터셋의 크기와 다양성의 중요성을 설명하세요.
    - 크기: 훈련 데이터의 크기와 다양성 덕분에 모델이 범용 지식을 필요로 하는 작업을 포함해 다양한 작업을 잘 수행할 수 있다.
    - 다양성: 다양한 주제의 텍스트는 모델이 언어의 구문, 문법, 맥락에 대한 깊은 이해를 발전시킬 수 있게 해준다.

3. 대규모 언어 모델에서 '토큰화'의 개념은 무엇인가요?
   - 모델이 텍스트를 처리할 수 있는 단위로 분활 하는 과정이다.
   - 단어, 부분 단어, 문자 등 다양한 수준에서 토큰화가 이루어진다.

4. 대규모 언어 모델의 '사전훈련' 개념과 중요성을 설명하세요.
   - 다양한 주제가 있는 대규모 데이터셋에서 사전훈련을 통해 언어에 대한 일반적인 이해를 구축한다.
   - 이렇게 준비된 사전훈련된 모델을 통해 특정 작업에 맞게 미세튜닝 할 수 있다.

5. 대규모 언어 모델의 맥락에서 미세 튜닝의 개념과 장점을 설명하세요.
   - 미세튜닝은 작업에 특화된 소규모 데이터셋에서 사전 훈련된 대규모 언어모델을 추가적으로 훈련한다.
   - 모델이 사전 훈련에서 배운 일반 지식을 활용하여 텍스트 요약이나 질문 답변과 같은 특정 작업을 잘 수행할수 있게 한다.

### Section 2-6 GPT 구조 자세히 살펴보기 <a class="anchor" id="section2-6"></a>
1. GPT 모델이 훈련되는 주요 작업은 무엇인가요? 번역과 같은 다른 작업을 수행하는 모델의 능력과 어떤 관련이 있나요?
   - GPT 모델은 다음 단어 예측 작업으로 훈련됩니다.
   - 이 작업을 통해 언어 사이의 패턴과 구조를 학습할 수 있으며, 이를 바탕으으로 번역작업을 위해 훈련되어있지 않더라도 번역 작업을 수행할 수 있는 능력을 갖추게 됩니다.

2. GPT 모델의 맥락에서 자기 지도 학습의 개념을 설명하세요.
   - GPT 모델은 다음 단어 예측으로 셀프 레이블링 방식의 자기 지도 학습을 수행한다.
   - 모델이 직접 레이블을 생성하는 방식으로, 대규모의 레이블이 없는 텍스트 데이터에서 훈련할 수 있다.

3. GPT 모델이 원본 트랜스포머 구조와 어떻게 다른가요? 이 차이가 의미하는 것은 무엇인가요?
    - GPT 모델은 원본 트랜스포머 구조에서 디코더 부분만 사용한다.
    - 이는 GPT 모델이 주로 텍스트 생성 작업에 초점을 맞추고 있음을 의미한다.

5. GPT 모델을 자기 회귀 모델로 간주하는 의미는 무엇인가요?
    - 한 번에 하나의 단어씩 예측하여 텍스트를 생성하고, 이전에 생성된 단어를 다음 단어 예측에 사용한다는 의미이다.

### Section 2-7 대규모 언어 모델 만들기 <a class="anchor" id="section2-7"></a>
1. 밑바닥부터 대규모 언어 모델을 만들기 위한 세 가지 단계는 무엇인가요?
   - 데이터 수집 및 전처리 --> LLM을 사전훈련하여 파운데이션 모델 구축 --> 파운데이션 모델을 미세튜닝 하여 맞춤형 LLM 생성

2. LLM에 사용되는 트랜스포머 구조의 핵심 아이디어는 무엇인가요?
    - 셀프 어텐션 메커니즘을 사용하여 입력 시퀀스의 서로 다른 부분에 선택적으로 주의를 기울이는 능력
    - 이를 통해 모델은 긴 문맥을 이해하고, 복잡한 언어 패턴을 학습할 수 있다.

3. GPT-3와 같은 LLM을 사전 훈련하는 데 사용되는 주요 작업은 무엇인가요?
   - 다음 단어 예측 작업
   - 모델이 주어진 단어 시퀀스에서 다음에 올 단어를 예측하도록 훈련한다.

4. LLM의 창발적 속성 개념에 대해 설명하세요.
    - 대규모 언어 모델이 훈련 데이터에서 명시적으로 학습하지 않은 새로운 능력을 나타내는 현상
    - 예: GPT-3가 번역 작업에 대해 훈련되지 않았음에도 불구하고 번역 작업을 수행할 수 있는 능력

5. 사전 훈련된 LLM을 미세 튜닝하는 것이 특정 작업에 도움이 되는 이유는 무엇인가요?
   - 특정 도메인이나 작업에서 사용되는 전문 용어와 문맥을 더 잘 이해할 수 있다.
   - 모델이 사전 훈련에서 배운 일반 지식을 활용하여 특정 작업에 적응할 수 있다.
