## Chapter 2-1 여러가지 미세튜닝 방법 <a class="anchor" id="chapter2-1"></a>
1. 언어 모델을 미세 튜닝하는 데 가장 널리 사용되는 두 가지 방법은 무엇인가요?
   - 지시 미세튜닝
      - 모델이 사용자의 구체적인 지시를 이해하고 이를 기반으로 응답을 생성한다.
      - 복잡한 사용자의 지시를 기반으로 다양한 작업을 처리할 수 있지만, 대규모 데이터셋과 계산 자원이 필요하다.
   - 분류 미세튜닝
      - 레이블이 있는 데이터셋을 사용하여 모델이 특정 클래스에 대한 예측을 수행하도록 한다.
      - 데이터와 계산 자원이 비교적 적게 필요하지만, 모델이 훈련된 특정 클래스에 대해서만 잘 작동한다.

2. 지시 미세 튜닝의 목적을 설명하고 예를 들어보세요.
   - 목적: 모델이 사용자의 구체적인 지시를 이해하고 이를 기반으로 응답을 생성한다.
   - 예: "이 문장을 프랑스어로 번역해줘."라는 지시를 받았을 때, 모델이 해당 문장을 프랑스어로 번역해서 응답한다.

3. 문류 미세 튜닝의 목적을 설명하고 예를 들어보세요.
   - 목적: 레이블이 있는 데이터셋을 사용하여 모델이 특정 클래스에 대한 예측을 수행하도록 한다.
   - 예: 영화 리뷰 데이터를 사용하여 긍정적인 리뷰와 부정적인 리뷰를 분류하는 모델을 훈련시킨다.

4. 분류 미세 튜닝된 모델의 주요 제약 사항은 무엇인가요?
   - 모델이 훈련된 특정 클래스에 대해서만 잘 작동한다.
   - 새로운 클래스나 도메인에 대한 일반화 능력이 제한적일 수 있다.
   - 레이블이 있는 데이터셋이 필요하며, 이는 수집과 라벨링에 비용이 발생할 수 있다.

5. 지시 미세 튜닝된 모델과 분류 미세튜닝된 모델의 유연성을 비교해 보세요.
    - 지시 미세 튜닝된 모델은 다양한 사용자의 지시를 이해하고 처리할 수 있는 유연성을 가지고 있다. 복잡한 작업을 수행할 수 있지만, 대규모 데이터셋과 계산 자원이 필요하다.
    - 분류 미세 튜닝된 모델은 특정 클래스에 대한 예측을 수행하는 데 특화되어 있다. 데이터와 계산 자원이 비교적 적게 필요하지만, 모델이 훈련된 특정 클래스에 대해서만 잘 작동한다.

6. 지시 미세 튜닝이 선호되는 경우는 언제인가요?
    - 사용자의 다양한 지시를 이해하고 처리해야 하는 경우
    - 복잡한 작업을 수행해야 하는 경우
    - 대규모 데이터셋과 계산 자원이 충분히 있는 경우

## Chapter 2-2 데이터셋 준비 <a class="anchor" id="chapter2-2"></a>

   - 형태: tab으로 구분되는 텍스트 파일로, 스펨/스펨아님 레이블과 해당 메시지 텍스트가 들어있다.

2. '스펨'과 '스펨 아님' 메시지의 개수가 동일하도록 언더샘플링 하는 이유가 무엇인가요?
   - 불균형한 클래스 분포를 해결하기 위해서이다. 
   - 만약 한 클래스가 다른 클래스에 비해 훨씬 많다면, 모델이 다수 클래스에 편향될 수 있다. 
   - 언더샘플링을 통해 각 클래스의 개수를 동일하게 맞추면, 모델이 두 클래스를 균등하게 학습할 수 있다.

   - "ham"은 0으로, "spam"은 1로 매핑하여 정수 클래스 레이블로 변환한다.
   - balanced_df["Label"].map({"ham": 0, "spam": 1})

## Chapter 2-5 분류 헤드 추가하기 <a class="anchor" id="chapter2-5"></a>
1. 분류 미세 튜닝을 위해 사전 훈련된 LLM의 출력층을 수정해야 하는 이유는 무엇이가요?
   - 사전 훈련된 LLM은 일반적으로 다음 단어를 예측하는 데 최적화되어 있으며, 분류 작업에 적합한 출력층이 아니다.
   - 따라서, 분류 작업에 맞게 출력층을 수정하여 모델이 특정 클래스에 대한 예측을 수행할 수 있도록 해야 한다.

2. 분류 작업에서 클래스 개수와 동일하게 출력 노드의 개수를 지정해야 하는 이유는 무엇인가요?
   - 각 출력 노드는 특정 클래스에 대한 예측을 나타내므로, 클래스 개수와 동일한 출력 노드 개수를 가져야 한다.
   - 이렇게 하면 모델이 각 클래스에 대한 확률을 출력할 수 있으며, 최종적으로 가장 높은 확률을 가진 클래스를 선택할 수 있다.

3. 새로운 작업에 사전 훈련된 LLM의 마지막 층만 미세 튜닝해도 충분한 이유는 무엇인가요?
   - 신경망 모델에서 앞쪽 층은 일반적으로 다양한 작업과 데이터셋에 적용할 수 있는 기본적인 언어 구조와 의미를 감지한다.
    - 마지막 층은 특정 작업에 맞게 조정되므로, 새로운 작업에 맞게 마지막 층만 미세 튜닝해도 충분하다.

4. 미세 튜닝을 위해 사전 훈련된 LLM의 층을 동결하고 해제하는 과정을 설명하세요.
   - 동결: 모델의 모든 파라미터의 requires_grad 속성을 False로 설정하여, 해당 파라미터들이 미세 튜닝 동안 업데이트되지 않도록 한다.
   - 해제
      - 마지막 트랜스 포머 블록과 정규화 층의 requires_grad 속성을 True로 설정하여, 이들 파라미터들이 미세 튜닝 동안 업데이트되도록 한다.
      - 분류 해드 층을 이진 분류에 맞게 생성 후 모델의 out_head에 할당한다. 층의 기본 파라미터는 requires_grad가 True로 설정된다.

5. 코잘 어텐션 마스크를 사용하는 분류 작업에서 시퀸스의 마지막 토큰에 가장 많은 정보가 들어 있다고 가정하는 이유는 무엇인가요?
   - 코찰 어텐션은 토큰의 현제 위치와 이전 토큰을 참조하므로 , 시퀸스의 마지막 토큰은 전체 시퀸스의 정보를 요약하는 역할을 한다.

6. GPT와 같은 모델을 분류 작업에 맞게 튜닝할 때, 마지막 출력 토큰에 집중하는 것이 중요한 이유를 설명하세요.
    - 마지막 출력 토큰은 전체 시퀸스의 정보를 요약하는 역할을 하므로, 분류 작업에서 중요한 특징을 포함하고 있다.
    - 따라서, 마지막 출력 토큰에 집중함으로써 모델이 더 정확한 분류를 수행할 수 있다.

## Chapter 2-6 분류 손실과 정확도 계산하기 <a class="anchor" id="chapter2-6"></a>
1. 모델의 출력을 스팸 분류를 위한 클래스 레이블 예측으로 변환하는 방법을 설명하세요.
   - 모델의 출력층을 스팸, 스팸아님 클래스를 구분하도록 설계된 분류 헤드로 변경한다.

2. 이 경우 softmax 함수의 목적은 무엇인가요? 이 함수가 선택 사항인 이유는 무엇인가요?
   - 목적: 모델의 출력 값을 확률 분포로 변환하여 각 클래스에 대한 예측 확률을 제공한다.
   - 선택 사항인 이유: 가장 큰 값의 인덱스를 선택하는 argmax 연산은 softmax 함수의 출력에 영향을 받지 않기 때문이다.

3. cal_accuracy_loader 함수의 기능과 이를 사용해 분류 정확도를 계산하는 방법을 설명하세요.
   - 기능: 입력 받은 배치 크기만큼 데이터 로더에서 데이터를 반복하여 모델의 출력 로짓을 얻는다.
   - 정확도 계산 방법
      - 마지막 출력 로짓에 대해 argmax 연산을 수행하여 예측된 클래스 레이블을 얻는다.
      - 입력된 샘를 겟수와 예측된 레이블과 실제 레이블을 비교하여 정확도를 계산한다.

4. 크로스 엔트로비 손실을 분류 정확도를 최대화하기 위해 대리자로 사용하는 이유는 무엇인가요?
   - 크로스 엔트로피 손실은 모델의 예측 확률과 실제 레이블 간의 차이를 측정하는 데 효과적이다.
   - 이 손실 함수를 최소화함으로써, 모델이 더 정확한 예측을 하도록 유도할 수 있다.
   - 따라서, 분류 정확도를 최대화하기 위한 대리 지표로 사용된다.

5. 분류를 위해 사용하는 cal_loss_batch와 언어 모델링을 위해 사용하는 것 사이의 차이점을 설명하세요.
   - cal_loss_batch 함수는 분류 작업에 특화된 손실 계산을 수행하며, 모델의 출력 로짓과 실제 레이블 간의 크로스 엔트로피 손실을 계산한다.
   - 반면, 언어 모델링을 위한 손실 계산은 일반적으로 다음 단어 예측을 위한 손실을 계산하며, 시퀸스 전체에 대한 손실을 고려한다.
   - 따라서, 두 함수는 목적과 계산 방식에서 차이가 있다.

## Chapter 2-8 LLM을 스팸 분류기로 사용하기 <a class="anchor" id="chapter2-8"></a>

   - 모델에 입력 시퀸스를 전달하여 출력 로짓을 얻는다.
   - 출력 로짓에 대해 argmax 연산을 수행하여 예측된 클래스 레이블을 얻는다.

    - 예측된 레이블과 실제 레이블을 비교하여 정확도를 계산한다.
