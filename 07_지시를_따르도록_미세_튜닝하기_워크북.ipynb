## Chapter 2-1 지시 미세튜닝 소개 <a class="anchor" id="chapter2-1"></a>
1. 사전 훈련된 LLM의 주요 기능은 무엇인가요?
    - 한 번에 한 단어씩 텍스를 생성하는 방법을 배우는 것입니다.
    - 이렇게 만들어진 사전 훈련된 LLM은 텍스트 완성능력이 있다.
    - "이 텍스트 문법을 고쳐 줘"와 같은 구체적인 명령을 잘 수행하지 못합니다.

2. 사전 훈련된 LLM이 종종 직면하는 어려움은 무엇인가요?
    - 구체적인 지시를 따르는 능력이 부족합니다.
    - 예를 들어, 문법 교정이나 어조 변환과 같은 작업을 잘 수행하지 못합니다.

3. 지시 미세 듀닝의 목적은 무엇인가요?
    - 사전 훈련된 LLM이 특정 지시를 더 잘 따르도록 만드는 것입니다.
    - 이를 통해 모델이 문법 교정, 어조 변환 등 다양한 작업을 더 효과적으로 수행할 수 있습니다.

4. 지시 미세 튜닝의 중요한 측면은 무엇인가요?
   - 지시-응답 쌍으로 구성된 데이터셋을 준비힌다.

## Chapter 2-2 지도 학습 지시 미세 튜닝을 위한 데이터셋 준비하기 <a class="anchor" id="chapter2-2"></a>
1. 사전 훈련된 LLM을 미세 튜닝하기 위한 지시 데이터셋의 역활은 무엇인가요?
    - 모델이 특정 지시를 더 잘 따르도록 돕습니다.
    - 지시-응답 쌍으로 구성되어 있습니다.

2. 이 절에서 사용하는 지시 데이터셋의 포멧을 설명하세요.
   - JSON 포멧을 사용합니다.
   - 각 데이터 항목은 지시, 입력, 응답의 세 가지 필드를 포함합니다.

3. 이 절에서 언급된 두 가지 프롬프트 스타일은 무엇인가요? 두 스타일은 어떻게 다른가요
   - 알파카: 지시, 입력, 응답의 세 가지 필드를 포함하는 JSON 포멧입니다.
   - Phi3: 지시와 응답의 두 가지 필드를 포함하는 JSON 포멧입니다.

4. format_input 함수의 역활은 무엇이며 어떻게 동작하나요?
    - format_input 함수는 지시와 입력을 결합하여 알파카 스타일 프롬프트를 생성합니다.
    - 입력이 존재하는 경우 지시와 입력을 함께 포맷하고, 입력이 없는 경우 지시만 포맷합니다.

5. 지시 데이터셋을 후련, 검증, 테스트 세트로 어떻게 나누나요?
   - 훈련, 테스트, 검증 비율에 따라 데이터를 나눈다.

## Chapter 2-3 훈련 배치 만들기 <a class="anchor" id="chapter2-3"></a>
1. 지시 미세 튜닝에서 사용자 정의 콜레이트 함수의 목적은 무엇인가요?
   - 배치의 샘플의 길이를 샘플 중 쵀대 길이에 맞추어 50256 패딩 토큰으로 패딩한다.
   - 입력과 타깃 시퀸스를 생성한다.
   - 타귓 시퀸스중 첫 번재 50256 패딩 토큰 이후의 모든 패딩 토큰을 -100 플레이스 홀더로 바꾼다.

2. 지시 미세 튜닝에서 사용자 정의 콜레이트 함수가 패딩을 어떻게 처리하나요?
   - 배치의 샘플의 길이를 샘플 중 쵀대 길이에 맞추어 50256 패딩 토큰으로 패딩한다.
   - 타킷 시퀸스중 첫 번재 50256 패딩 토큰 이후의 모든 패딩 토큰을 -100 플레이스 홀더로 바꾼다.

3. 지시 미세 튜닝에서 타킷 토큰 ID의 역활과 생성하는 방법을 설명하세요.
    - 타킷 토큰 ID는 모델이 예측해야 하는 목표 시퀸스를 나타냅니다.
    - 입력 시퀸스를 오른쪽으로 한 칸 이동시켜 생성합니다.

4. 타깃 토큰 ID에서 패딩 토큰을 플레이스홀더 값인 -100으로 바꾸는 이유는 무엇인가요?
    - 크로스엔트로피 손실 계산에서 패딩 토큰을 제외하기 위해서입니다.
    - 이렇게 하면 모델이 패딩 토큰에 대해 학습하지 않도록 할 수 있습니다.

5. 타깃 시퀸스에서 하나의 텍스트 종료 토큰을 남겨 놓는 이유는 무엇인가요?
    - LLM이 응답의 끝을 나타내는 텍스트 종료 토큰을 생성하는 방법을 학습할 수 있도록 하기 위해서입니다.
